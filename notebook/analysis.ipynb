```python
#11. Loans by karat
#After removing duplicates and converting purity to a karat number (916 = 22, 833 = 20, 750 = 18, 999 = 24), return karat and loans for each karat.
loans['purity'].unique()
```

```python
karat={'916':22,'833':20,'750':18,'999':24}

def to_karat(p):
    num=''.join(i for i in str(p) if i.isdigit())
    return karat.get(num,int(num))

loans['karat']=loans['purity'].apply(to_karat)
```

```python
loans['karat'].unique()
```

```python
loans.head()
```

```python
answer=loans.groupby('purity')['loan_id'].count().reset_index(name='loans')
```

```python
answer=answer.rename({'purity':'karat'})
```
